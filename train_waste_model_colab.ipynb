```python
from google.colab import files
uploaded = files.upload()
print(list(uploaded.keys()))
```

```python
import zipfile, glob, os
z = next(iter(uploaded.keys()))
with zipfile.ZipFile(z) as f: f.extractall('/content')
print(glob.glob('/content/**/dataset.yaml', recursive=True))
```

```python
from ultralytics import YOLO
DATA = glob.glob('/content/**/dataset.yaml', recursive=True)[0]
model = YOLO('yolo26n.pt')
model.train(data=DATA, epochs=50, imgsz=416, batch=8, project='/content/runs', name='waste_detection')
```

```python
results = model.val()
```
